```python
# 1. Show the next answer to label (question and answer only)
todo = [i for i in labels.items if i not in labels.state["first"]]
print(f"{len(todo)} left")
if todo:
    item = labels.items[todo[0]]
    print(item["id"], "\n\nQ:", item["question"], "\n\nA:", item["answer"])
```

```python
# 2. Label it: PASS with rule "pass", or FAIL with the rubric rule 1-7, and a one-line reason
labels.label(todo[0], "PASS", "pass", "<one line: what the answer says and what the policy says>")
```

```python
# 3. After all first labels: the flags, to read again (they never change a label)
print("\n".join(labels.flags()) or "no flags")
```

```python
# 4. At least 7 days later: the re-label sample, with the first labels hidden
for row in labels.to_relabel():
    print(row["id"], "\nQ:", row["question"], "\nA:", row["answer"], "\n")
```
